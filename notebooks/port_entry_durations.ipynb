# Port entry durations for programs that store them in U

`get_no_port_entry_duration_sessions` reads the port entry durations from `E` and falls back to `U` for programs such as `Probe Test Habit Training TTL`, where `E` is a single-line variable. The fallback caught `TypeError`, but `read_medpc_file` raises `ValueError` for that case, so running the dataset conversion without a cached `sessions_without_port_entry_durations.yaml` stopped at the first probe test. This notebook runs the check over every session in the dataset with the fix, then reconverts one probe test and compares its port entry intervals with the published file.

```python
from datetime import datetime
from pathlib import Path
import shutil

import h5py
import numpy as np
import pandas as pd

from lerner_lab_to_nwb.seiler_2024.seiler_2024_convert_dataset import (
    fp_to_nwb,
    opto_to_nwb,
    get_no_port_entry_duration_sessions,
    get_session_key_from_kwargs,
)
from lerner_lab_to_nwb.seiler_2024.seiler_2024_convert_session import session_to_nwb

data_dir_path = Path("/Volumes/T7/CatalystNeuro/Lerner/_Raw_Data_for_future_reference")
published_dir_path = Path("/Volumes/T7/CatalystNeuro/Lerner/000971")
output_dir_path = Path("/Volumes/T7/CatalystNeuro/Lerner/fix_audit/port_entry_durations")
if output_dir_path.exists():
    shutil.rmtree(output_dir_path)
output_dir_path.mkdir(parents=True)
```

```python
session_to_nwb_kwargs_per_session = fp_to_nwb(
    data_dir_path=data_dir_path, output_dir_path=output_dir_path, start_variable="Start Date", verbose=False
) + opto_to_nwb(data_dir_path=data_dir_path, output_dir_path=output_dir_path, start_variable="Start Date", verbose=False)
no_port_entry_duration_sessions = get_no_port_entry_duration_sessions(
    port_entry_file_path=output_dir_path / "sessions_without_port_entry_durations.yaml",
    session_to_nwb_args_per_session=session_to_nwb_kwargs_per_session,
)
print(f"{len(no_port_entry_duration_sessions)} of {len(session_to_nwb_kwargs_per_session)} sessions have no port entry durations")
```

```python
subject_id, start_datetime = "028.392", datetime(2020, 8, 3, 11, 23, 3)
(session_to_nwb_kwargs,) = [
    session_to_nwb_kwargs
    for session_to_nwb_kwargs in session_to_nwb_kwargs_per_session
    if session_to_nwb_kwargs["subject_id"] == subject_id
    and session_to_nwb_kwargs["session_conditions"].get("Start Date") == start_datetime.strftime("%m/%d/%y")
    and session_to_nwb_kwargs["session_conditions"].get("Start Time") == start_datetime.strftime("%H:%M:%S")
]
if get_session_key_from_kwargs(session_to_nwb_kwargs) in no_port_entry_duration_sessions:
    session_to_nwb_kwargs["has_port_entry_durations"] = False
session_to_nwb(**session_to_nwb_kwargs)
```

```python
session_stamp = start_datetime.isoformat().replace(":", "-")
(published_file_path,) = (published_dir_path / "sub-028-392").glob(f"*{session_stamp}*.nwb")
(new_file_path,) = output_dir_path.glob(f"sub-{subject_id}_*{session_stamp}*.nwb")


def read_behavior_datasets(file_path: Path) -> dict[str, np.ndarray]:
    datasets = {}
    with h5py.File(file_path, "r") as file:
        file["processing/behavior"].visititems(
            lambda name, item: datasets.__setitem__(name, item[()]) if isinstance(item, h5py.Dataset) else None
        )
    return datasets


published_datasets = read_behavior_datasets(published_file_path)
new_datasets = read_behavior_datasets(new_file_path)
pd.DataFrame(
    [
        dict(
            dataset=name,
            published_length=len(published_datasets.get(name, [])),
            new_length=len(new_datasets.get(name, [])),
            equal=name in published_datasets
            and name in new_datasets
            and np.array_equal(published_datasets[name], new_datasets[name]),
        )
        for name in sorted(set(published_datasets) | set(new_datasets))
    ]
)
```

This probe test has no port entry durations in `U`, so it is converted with port entry times only, as in the published file.
